## 5. Hands-on 3: Simulating drug inhibition in prostate cell-line-specific Boolean models
### 5.1 Simulating prostate cell-line-specific Boolean models

In this article, Montagud et al. generated multiple personalized versions of the model, corresponding to patients or cell lines. Here, we're going to work with a subset of them for simplicity. 
In the next block, we build a dictionnary of models, with an identifier as key, and the Simulation object as value

````python
patients_models = {}
bnd_files = sorted(glob.glob("models/cell_lines/Montagud2022_Prostate_Cancer_mut_RNA_*.bnd"))
for bnd in bnd_files:
    cfg = os.path.splitext(bnd)[0] + '.cfg'
    m = re.search(r"_mut_RNA_(\d+)", bnd)
    if m:
        idx = m.group(1)
    else:
        idx = os.path.splitext(os.path.basename(bnd))[0]
    patients_models[f'model_{idx}'] = maboss.load(bnd, cfg)
patients_models
````
